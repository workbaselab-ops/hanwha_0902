p91 : 주요 매개변수를 invoke() 함수로 응답 출력하기

```python
from dotenv import load_dotenv  # .env 환경변수 사용
from langchain_openai import ChatOpenAI
```

```python
# .env에 저장된 한경변수 불러오기
load_dotenv()
```

```python
# ChatOpenAI 객체 생성
llm = ChatOpenAI(                # 객체생성
    temperature=0.1,             # 창의성 (0.0~20.)
    model_name="gpt-5-mini",     # 모델명
)
```

```python
question = "대한민국의 수도는 어디야?"
```

```python
print(f"[답변]: {llm.invoke(question)}")  #답변
```

```python
response = llm.invoke(question)  #답변
response
```

```python
response.content
```

```python
response.response_metadata
```

p92 : logprob 활성화하기

```python
# ChatOpenAI 객체 생성
llm_with_logprob = ChatOpenAI(   # 객체생성
    temperature=0.1,             # 창의성 (0.0~20.)
    max_tokens = 2048,           # 최대 토큰 수
    model_name="gpt-4o-mini"      # 모델명
).bind(logprobs=True)
```

```python
question = "대한민국의 수도는 어디인가요?"
```

```python
response = llm_with_logprob.invoke(question)
response.response_metadata
```

p93 : stream() 함수로 스트리밍 출력하기

```python
# 스트림 방식으로 질의.answer에 스트리밍 답변의 결과를 받는다.
answer = llm.stream("대한민국의 아름다운 관광지 10곳과 주소를 알려줘!")
# 마지막 줄에는 주석 달지 말기
for token in answer:
    print(token.content, end="", flush=True)   
```

```python
answer = llm.stream("대한민국의 아름다운 관광지 10곳과 주소를 알려줘!")

final_answer = ""
for token in answer:
    final_answer +=token.content

print(final_answer)
```

```python
from langchain_teddynote.messages import stream_response

answer = llm.stream("대한민국의 아름다운 관광지 10곳과 주소를 알려주세요!")
stream_response(answer)
```

```python
answer = llm.stream("대한민국의 아름다운 관광지 10곳과 주소를 알려주세요!")
final_answer2 = stream_response(answer, return_output=True)
print(final_answer2)
```
