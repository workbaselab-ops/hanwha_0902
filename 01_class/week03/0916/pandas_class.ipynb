Locate Row
Pandas use the loc attribute to return one or more specified row(s)



````python
#refer to the row -> ` locreturn` 속성을 사용하여 하나 이상의 지정된 행 반
print(df1.loc[0])  # 0번째 행을 반환
````

````python
#use a list of indexes:
print(df1.loc[[0, 1]])   # 0번째 행과 1번째 행을 반환
````

**Pandas의 notna()**는 데이터프레임(DataFrame)이나 시리즈(Series) 내의 **각 값이 결측치(Null, NaN, None)가 아닌지 검사하는 함수**입니다.

동작 방식: 각 요소가 정상적인 데이터이면 True, 결측치(NaN 등)이면 False를 반환합니다.

**주요 용도**: 데이터 전처리 시 **결측치를 제외한 유효한 데이터만 필터링할 때** 주로 사용됩니다.

````python
age_no_na = titanic[titanic["Age"].notna()]
age_no_na.head()
````

````python
age_no_na.shape
````

````python
adult_names = titanic.loc[titanic["Age"] > 35, "Name"]

adult_names.head()
````

````python
titanic.iloc[9:25, 2:5]  # rows 10 till 25 and columns 3 to 5.
````
