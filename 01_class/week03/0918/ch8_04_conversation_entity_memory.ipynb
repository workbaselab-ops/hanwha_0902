# ConversationEntityMemory

엔티티 메모리는 대화에서 특정 엔티티에 대한 주어진 사실을 기억합니다.

엔티티 메모리는 엔티티에 대한 정보를 추출하고(LLM 사용) 시간이 지남에 따라 해당 엔티티에 대한 지식을 축적합니다(역시 LLM 사용).

```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
from langchain_classic.chains import ConversationChain
from langchain_classic.memory import ConversationEntityMemory
from langchain_classic.memory.prompt import ENTITY_MEMORY_CONVERSATION_TEMPLATE
```

Entity 메모리를 효과적으로 사용하기 위하여, 제공되는 프롬프트를 사용합니다.

```python
# Entity Memory를 사용하는 프롬프트 내용 출력
print(ENTITY_MEMORY_CONVERSATION_TEMPLATE.template)
```

```python
# LLM 생성
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

#  ConversationChain 생성
conversation = ConversationChain(
    llm = llm,
    prompt = ENTITY_MEMORY_CONVERSATION_TEMPLATE,
    memory = ConversationEntityMemory(llm=llm),
)
```

대화를 시작합니다.

입력한 대화를 바탕으로 `ConversationEntityMemory` 는 주요 Entity 정보를 별도로 저장합니다.

```python
conversation.predict(
    input="희영은 K-뉴딜 한화 아카데미를 수강중에있습니다."
    "6년차 PM으로 일한 경험이 있습니다."
    "수업과 동시에 취업을 하기 위해 지원서와 포트폴리오를 정리하고있습니다."
)
```

Entity는 `memory.entity_store.store` 에서 확인할 수 있습니다.

```python
# entity memory 출력
conversation.memory.entity_store.store
```
