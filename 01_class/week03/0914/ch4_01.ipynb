
### 채팅모델 사용해보기

```python
# 코드실행에 필요한 라이브러리를 상단위치에 선언
from langchain_openai import ChatOpenAI
```

```python
# 챗모델 객체 생성 = 클래스 생성
llm = ChatOpenAI(
    temperature=0.1,      # 창의성(0.0~2.0)
    model="gpt-4o-mini"   # 모델명
)
```

```python
# 질문
question = "미국의 수도는 어디인가요?"
```

답변의 형식(AI Message)

```python
# 응답

# reponse를 쓰는 이유는 뒤 함수를 변수로 만든것
# -> 아래에서도 question을 썼을때 계속 response만 쳤을 떄 응답받기위함
response = llm.invoke(question)
```

```python
# 응답 호출
response
```

```python
# 답변만 받을 때
response.content
```

```python
response.response_metadata
```

### LogProb 활성화

- 모델이 단어(토큰)를 생성할 때 "이 단어를 얼마나 확신을 가지고 선택했는지"에 대한 확률값을 얻을 수 있다.

모델의 확신도(Confidence Level) : 0에 가까울수록 정답에 가까운 확신을 가진다

```python
# 객체(obj) 생성
llm_withlogprob = ChatOpenAI(
    temperature=0.1,
    max_completion_tokens=2048,
    model="gpt-4o-mini"
).bind(logprobs=True)
```

```python
question2 = "인도의 수도는 어디인가요?"

response = llm_withlogprob.invoke(question2)
```

```python
response
```

```python
response.response_metadata
```

```python
response.response_metadata["logprobs"]
```

### Stream 실시간 출력 방식

스트리밍 옵션은 질의에 대한 답변을 한 토큰씩 받아와서 화면에 실시간으로 받을 때 유용합니다.

```python
# answer 에 스트리밍 답변의 결과를 받습니다. (실시간 출력)
answer = llm.stream("LangChain 8단계 구성과 개념에 대해 알려줘!")
```

출력 버퍼(buffer)
: 중간에 잠깐 저장해 두는 공간

flush
: "버퍼에 남아 있는 출력을 지금 바로 화면으로 보내라"는 의미
-> flush=False: 결과가 버퍼에 저장되었다가, 한번에 출력.
-> flush=True : 결과가 생성되는대로 바로바로 출력.

특히, 실시간으로 뭔가 출력해야 할 때 유용
-> LangChain의 stream()이 대표적인 예

end=""
-> "출력하고 줄바꿈하지 마."라는 뜻

```python
answer = llm.stream("LangChain 8단계 구성과 개념에 대해 알려줘!")
```

```python
# 화면에 실시간 출력
for token in answer:
    print(token.content, end="", flush=True)
```

+=
: 받은 토큰을 최종 문자열에 계속 저장

+= token.content
: 토큰으로 조각조각 받은 내용을 계속 이어 붙이라

```python
#스트리밍 답변결과 변수에 저장
answer = llm.stream("LangChain 8단계 구성과 개념에 대해 알려줘!")

final_answer = ""
for token in answer :
    final_answer += token.content
```

```python
final_answer
```

```python
print(final_answer)
```

```python
# 화면에 실시간 출력 + 스트리밍 답변결과 변수에 저장
answer = llm.stream("LangChain 8단계 구성과 개념에 대해 알려줘!")

for token in answer:
    print(token.content, end="", flush=True)
    final_answer += token.content
```
